# 2. Data Cleaning

In this section, the dataset will be cleaned and validated to ensure data quality and consistency for further analysis.

```python
# Import required libraries

import pandas as pd
import numpy as np
```

```python
# Load Dataset

df = pd.read_csv("../data/raw/Palo Alto Networks.csv")
```

```python
# Check dataset shape

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Check missing values

missing_values = df.isnull().sum()

missing_values[missing_values > 0]
```

```python
# Check duplicate records

duplicate_count = df.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

```python
# Check data types

df.dtypes
```

```python
# Check categorical values

categorical_columns = [
    "Attrition",
    "BusinessTravel",
    "Department",
    "Gender",
    "JobRole",
    "MaritalStatus",
    "OverTime"
]

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
# Validate ordinal columns

ordinal_columns = [
    "EnvironmentSatisfaction",
    "JobInvolvement",
    "JobSatisfaction",
    "RelationshipSatisfaction",
    "WorkLifeBalance"
]

for column in ordinal_columns:
    print(f"{column}: {sorted(df[column].unique())}")
```

```python
# Check minimum and maximum values of numerical columns

df.describe().T[["min", "max"]]
```

```python
# Remove duplicate records

df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
# Reset index after cleaning

df = df.reset_index(drop=True)
```

```python
# Save cleaned dataset

output_path = "../data/processed/employee_engagement_dataset.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully.")
print("File:", output_path)
```

## Observation

The dataset was successfully validated and cleaned. No missing values or duplicate records were identified, and the key categorical and ordinal variables were found to be consistent with the project requirements.

The cleaned dataset has been saved in the processed data directory for further exploratory and feature engineering analysis.
